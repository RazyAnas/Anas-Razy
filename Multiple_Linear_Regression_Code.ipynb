#Making our own linear regression class

```python
class MeraLR():
  def __init__(self):
    self.coef_ = None
    self.intercept_ = None

  def fit(self, X_train, y_train):
    # using np.insert we can add a column directly to matrix
    X_train = np.insert(X_train, 0, 1, axis=1) # to matrix, column number, what value, at what axis

    # calculate coefficitent
    # to calculate inverse of a matrix we use np.linalg.inv
    betas = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
    self.intercept_ = betas[0]
    self.coef_ = betas[1:]
    print(betas)

  def predict(self, X_test):
    y_pred = np.dot(X_test, self.coef_) + self.intercept_
    return y_pred
```

```python
lr = MeraLR()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.pred = lr.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```
